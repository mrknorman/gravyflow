# Notebook 00: Installation Guide

## 1. Install GravyFlow
GravyFlow needs Python 3.13 or later. Install it from the repository, into a new environment, with the extras you need
(`realdata` for real detector data, `viz` for plots and validation, `notebooks` for these notebooks):

```bash
python3.13 -m venv gravyflow-env
source gravyflow-env/bin/activate
pip install "gravyflow[notebooks] @ git+https://github.com/mrknorman/gravyflow"
```

Add the `cuda` extra on Linux with NVIDIA GPUs. To work on GravyFlow itself, install a clone in editable mode:

```bash
git clone https://github.com/mrknorman/gravyflow
cd gravyflow
pip install -e ".[test,notebooks]"
```

## 2. Choose Keras's JAX backend
GravyFlow runs on Keras 3's JAX backend, and `import gravyflow` changes nothing in your process, so choose the backend
before Keras is imported, in the shell:

```bash
export KERAS_BACKEND=jax
```

or at the top of a notebook, before importing Keras or GravyFlow (as every notebook here does):

```python
import os
os.environ.setdefault("KERAS_BACKEND", "jax")
import gravyflow as gf
```

## 4. Public data sources

Real noise and glitches use public data only; no credentials are needed:

- Segments and observing-run bounds come from the GWOSC timeline and run API (https://gwosc.org). The default selection is `DATA` with the `BURST_CAT1` and `BURST_CAT2` vetoes applied, without CBC, burst, stochastic or detchar hardware injections (continuous-wave injections, which run throughout, are kept).
- Glitches come from the Gravity Spy machine-learning classifications: Glanzer et al. 2023, *Class. Quantum Grav.* 40 065004, released on Zenodo (doi:10.5281/zenodo.5649212, v1.0.0, CC-BY-4.0). Please cite both if you use them. Each file is downloaded on first use and cached in `~/.gravyflow/cache` (override with `GRAVYFLOW_CACHE_DIR`).

## 5. Test Gravyflow (optional)

GravyFlow includes PyTest for testing its functionality. To run tests:

```bash
pytest gravyflow
```

Note: Tests may fail due to unavailable GPU memory if GPUs are currently under heavy use.
